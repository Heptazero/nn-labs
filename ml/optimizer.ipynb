- 本次目标函数：Rosenbrock 香蕉函数

```
def f(x, y):
    return (1 - x)**2 + 100 * (y - x**2)**2

def grad_f(x, y):       # 梯度，一个向量场
    df_dx = -2*(1 - x) - 400*x*(y - x**2)
    df_dy = 200*(y - x**2)
    return np.array([df_dx, df_dy])

def hess_f(x, y):          # 牛顿法要用
    d2x  = 2 - 400*y + 1200*x**2
    dxdy = -400*x
    d2y  = 200
    return np.array([[d2x, dxdy], [dxdy, d2y]])
```

- 优化器代码 = 拿着上面这三个函数当"工具",写一个循环,一步步挪动点的位置:
只要代码里出现了 x_{k+1} = x_k - ... 这种迭代更新,就已经是优化器代码了。
- 优化器 = 你选择用什么方式从山顶走下去（是迈小碎步、拄拐杖探路，还是直接往谷底跳）。

```

def gradient_descent(grad_f, x0, lr, n_iter,verbose=True):  # grad_f是传的一个函数
  # 初始化
    x = x0.astype(np.float64)  # 确保是浮点数
    history = [x.copy()]  # 记录轨迹
    for i in range(n_iter):
      grad = grad_f(x[0], x[1])
      x = x - lr * grad # x_n+1=x_n - 学习率(lr)x梯度(grad_f)
      history.append(x.copy())
      if verbose and (i + 1) % 100 == 0:
        print(f"迭代 {i+1}: x = {x[0]:.6f}, y = {x[1]:.6f}")
    return x, history

def newton(grad_f, hess_f, x0, n_iter, tol=1e-8,verbose=False):
  x = x0.astype(np.float64)
  history = [x.copy()]  # 记录轨迹
  for i in range(n_iter):
    grad = grad_f(x[0], x[1])
    if np.linalg.norm(grad) < tol:          # 收敛判定
      if verbose:
        print(f"第{i}步收敛,梯度范数={np.linalg.norm(grad):.2e}")
      break
    H = hess_f(x[0], x[1])  # diff1:输入参数多了个 hess_f：因为牛顿法需要二阶导信息，所以必须把海森矩阵传进去
    step = np.linalg.solve(H, grad)   #diff2:牛顿法的步长是数学算出来的
    # 解线性方程组,别用 np.linalg.inv(H) @ grad，solve 比 inv 高明在三个地方
    x = x - step
    history.append(x.copy())
    if verbose and (i + 1) % 10 == 0:
      print(f"迭代 {i+1}: x = {x[0]:.6f}, y = {x[1]:.6f}")
  return history


def coordinate_descent(x0, n_iter, tol=1e-8, verbose=False):
  x = x0.astype(np.float64)
  history = [x.copy()]
  for i in range(n_iter):
    for j in range(len(x)):          # 新增的内层循环:一个一个坐标来
      x[j] = ???                     # 固定其他坐标,只对x[j]求最优
    history.append(x.copy())
    # 收敛判定...
  return history
```
